```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Mall_Customers.csv")

print(df.head())
print(df.info())
print(df.isnull().sum())
```

```python
from sklearn.preprocessing import StandardScaler

X = df[['Annual Income (k$)',
        'Spending Score (1-100)']]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.cluster import KMeans

inertias = []

for k in range(1,11):
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    model.fit(X_scaled)
    inertias.append(model.inertia_)

plt.plot(range(1,11), inertias, marker = 'o')
plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.show()
```

```python
k = 5 # Adjuct based on the Elbow Method

kmeans = KMeans(
    n_clusters=k,
    random_state=42,
    n_init=10
)

kmeans_labels = kmeans.fit_predict(X_scaled)

plt.scatter(
    X_scaled[:,0],
    X_scaled[:,1],
    c=kmeans_labels,
    cmap='viridis'
)

plt.scatter(
    kmeans.cluster_centers_[:,0],
    kmeans.cluster_centers_[:,1],
    c='red',
    marker='X',
    s=200,
    label='Centroids'
)

plt.xlabel("Standardized Annual Income")
plt.ylabel("Standardized Spending Score")
plt.title("K-Means Clustering")
plt.legend()
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering

hierarchial = AgglomerativeClustering(
    n_clusters=k,
    linkage='ward'
)

hierarchial_labels = (
    hierarchial.fit_predict(X_scaled)
)

plt.scatter(
    X_scaled[:,0],
    X_scaled[:,1],
    c=hierarchial_labels,
    cmap='viridis'
)

plt.xlabel("Standardized Annual Income")
plt.ylabel("Standardized Spending Score")
plt.title("Agglomerative Hierarchial Clustering")
plt.show()
```

```python
from scipy.cluster.hierarchy import (
    linkage, dendrogram
)

linked = linkage(X_scaled, method = 'ward')

plt.figure(figsize=(12,5))
dendrogram(linked)
plt.title("Hierarchial Clustering Dendrogram")
plt.xlabel("Customers")
plt.ylabel("Linkage Distance")
plt.show()
```

```python
from sklearn.metrics import silhouette_score
import numpy as np

kmeans_sil = silhouette_score(
    X_scaled, kmeans_labels
)

hierarchial_sil = silhouette_score(
    X_scaled, hierarchial_labels
)

def calculate_inertia(X, labels):
    total=0
    for cluster in np.unique(labels):
        points = X[labels == cluster]
        centroid = points.mean(axis=0)
        total += np.sum(
            (points - centroid)**2
        )
    return total

results = pd.DataFrame({
    "Algorithm": [
        "K-Means", "Agglomerative"
    ],
    "Inertia":[
        kmeans.inertia_,
        calculate_inertia(
            X_scaled, hierarchial_labels
        )
    ],
    "Silhouette Score": [
        kmeans_sil, hierarchial_sil
    ]
})

print(results)

```
